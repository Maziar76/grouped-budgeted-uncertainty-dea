## Before running

1. Install dependencies: `python -m pip install -r requirements.txt`.
2. Obtain the original supplementary workbook from the source publication and place it at `data/1-s2.0-S2352340918314410-mmc2.xlsx`. The workbook is not redistributed.
3. Run the cells below in order from the repository root.

The random-partition experiment is computationally intensive and supports checkpoint-based resumption. Do not alter the country-region mapping or variable specification without checking the manuscript and original source documentation.

## Output and interpretation notes

- Main-analysis outputs are written under `results/GBU_DEA_outstanding_3x3_results/`.
- Random-partition outputs are written under `results/random_partition_1000_results/`.
- The validation script creates `results/GBU_DEA_reproducibility_outputs.zip` only after checking the expected record counts.
- Proposition 6 equality frequencies use solver-selected optimal intensity vectors; where the underlying LP has multiple optima, these diagnostics are representation-dependent.
- The random-partition quantile ranges are empirical Monte Carlo ranges, not bootstrap confidence intervals.
- The country-to-region mapping requires confirmation against the original supplementary documentation before the results are represented as independently verified.
